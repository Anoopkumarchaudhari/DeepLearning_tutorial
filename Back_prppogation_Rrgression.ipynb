```python
def param_init(layers_dense):
  np.random.seed(3)
  parameter={};
  L = len(layers_dense)
  for i in range(1,L):
    parameter['w'+str(i)]=np.ones((layers_dense[i-1],layers_dense[i]))*0.1
    parameter['b'+str(i)]=np.zeros((layers_dense[i],1))

  return parameter
```

```python
param_init([2,2,1])
```

```python
def linear_forward(x,w,b):
  z = np.dot(w.T,x)+b

  return z
```

```python
#main fUnction foe forward Prpogation
def L_layer_forward(X,parameter):
  A = X
  L = len(parameter)//2 # No. of layers in neural network
  for i in range(1,L+1):
    A_prev = A
    w1 = parameter['w'+str(i)]
    b1 = parameter['b'+str(i)]
    # print('A'+str(i-1)+': ',A_prev)
    # print('w'+str(i)+': ',w1)
    # print('b'+str(i)+': ',b1)
    # print('---'*15)

    A = linear_forward(A_prev,w1,b1)
    # print('A'+str(i)+': ',A)
    # print('***'*15)

  return A,A_prev
```

```python
def update_parameters(parameters,y,y_hat,A1,X):
  parameters['w2'][0][0] = parameters['w2'][0][0] + (0.001 * 2 * (y - y_hat)*A1[0][0])
  parameters['w2'][1][0] = parameters['w2'][1][0] + (0.001 * 2 * (y - y_hat)*A1[1][0])
  parameters['b2'][0][0] = parameters['w2'][1][0] + (0.001 * 2 * (y - y_hat))

  parameters['w1'][0][0] = parameters['w1'][0][0] + (0.001 * 2 * (y - y_hat)*parameters['w2'][0][0]*X[0][0])
  parameters['w1'][0][1] = parameters['w1'][0][1] + (0.001 * 2 * (y - y_hat)*parameters['w2'][0][0]*X[1][0])
  parameters['b1'][0][0] = parameters['b1'][0][0] + (0.001 * 2 * (y - y_hat)*parameters['w2'][0][0])

  parameters['w1'][1][0] = parameters['w1'][1][0] + (0.001 * 2 * (y - y_hat)*parameters['w2'][1][0]*X[0][0])
  parameters['w1'][1][1] = parameters['w1'][1][1] + (0.001 * 2 * (y - y_hat)*parameters['w2'][1][0]*X[1][0])
  parameters['b1'][1][0] = parameters['b1'][1][0] + (0.001 * 2 * (y - y_hat)*parameters['w2'][1][0])
```

```python
#epochs implimentation
parameter = param_init([2,2,1])
epochs = 10
for i in range(epochs):
  loss = []

  for j in range(df.shape[0]):
    X= df[['cgpa','resume_score']].values[j].reshape(2,1)
    y = df[['package']].values[j][0]

    y_hat,A1 = L_layer_forward(X,parameter)
    y_hat = y_hat[0][0]

    update_parameters(parameter,y,y_hat,A1,X)

    loss.append((y-y_hat)**2)

  print(f'epochs:{i}  loss:{np.array(loss).mean()}')

parameter
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(2,activation='linear',input_dim=2))
model.add(Dense(1,activation='linear'))
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
new_weight = [np.array([[0.1,0.1],[0.1,0.1]],dtype=np.float32),np.array([.0,.0],dtype=np.float32),
              np.array([[0.1],[0.1]],dtype=np.float32),np.array([.0],dtype=np.float32)]
```

```python
new_weight
```

```python
model.set_weights(new_weight)
```

```python
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss='mean_squared_error',optimizer=optimizer)
```

```python
X = df.drop(columns=['package']).values
y = df['package'].values
```

```python
model.fit(X,y,epochs=50,verbose=1,batch_size=1)
```
